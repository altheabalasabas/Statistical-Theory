```python

import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from math import sqrt

# 1. MULTIPLE REGRESSION EQUATION

# Given data
X1 = np.array([3,5,6,8,12,14])
X2 = np.array([16,10,7,4,3,2])
X3 = np.array([90,72,54,42,30,12])

# Independent variables
X = np.column_stack((X1, X2))

# Regression model
model = LinearRegression()
model.fit(X, X3)

# Coefficients
a = model.intercept_
b1, b2 = model.coef_

print("1. Multiple Regression Equation")
print(f"X3 = {a:.2f} + ({b1:.2f})X1 + ({b2:.2f})X2")


# 1b. ESTIMATE X3 when X1=10 and X2=6

x1_new = 10
x2_new = 6

predicted = model.predict([[x1_new, x2_new]])

print("\nEstimated X3 when X1=10 and X2=6:")
print(f"X3 = {predicted[0]:.2f}")

# 2. STANDARD ERROR OF ESTIMATE

# Predicted values
y_hat = model.predict(X)

# Residuals
residuals = X3 - y_hat

# Sum of squared errors
SSE = np.sum(residuals**2)

n = len(X3)
k = 2   # number of predictors

# Standard error
SEE = sqrt(SSE / (n - k - 1))

print("\n2. Standard Error of Estimate")
print(f"SEE = {SEE:.2f}")


# 3. COEFFICIENT OF MULTIPLE CORRELATION

# Total Sum of Squares
SST = np.sum((X3 - np.mean(X3))**2)

# Regression Sum of Squares
SSR = SST - SSE

# Multiple correlation coefficient
R = sqrt(SSR / SST)

print("\n3. Coefficient of Multiple Correlation")
print(f"R = {R:.4f}")


# 4. PARTIAL CORRELATION
# r12.3 = correlation between X1 and X2
# controlling X3

# Correlations
r12 = np.corrcoef(X1, X2)[0,1]
r13 = np.corrcoef(X1, X3)[0,1]
r23 = np.corrcoef(X2, X3)[0,1]

# Partial correlation formula
numerator = r12 - (r13 * r23)
denominator = sqrt((1 - r13**2) * (1 - r23**2))

r123 = numerator / denominator

print("\n4. Partial Correlation Coefficient")
print(f"r12.3 = {r123:.4f}")


# 5. REAL-WORLD MULTIPLE REGRESSION

# Housing data
price = np.array([165,200,225,180,202,250,275,300,155,230])

bedrooms = np.array([3,3,4,2,4,4,3,5,2,4])
baths = np.array([2,3,3,3,2,4,4,3,2,4])

# Independent variables
X_house = np.column_stack((bedrooms, baths))

# Regression model
house_model = LinearRegression()
house_model.fit(X_house, price)

# Coefficients
a2 = house_model.intercept_
b_bed, b_bath = house_model.coef_

print("\n5. Real-World Multiple Regression")
print(f"Price = {a2:.2f} + ({b_bed:.2f})Bedrooms + ({b_bath:.2f})Baths")

# Predict price for 5 bedrooms and 4 baths
pred_house = house_model.predict([[5,4]])

print("\nEstimated price for a house with 5 bedrooms and 4 baths:")
print(f"Price = {pred_house[0]:.2f}")
```
