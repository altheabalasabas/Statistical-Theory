```python
import statistics
import math

print("Question 1:")
males = [
12,4,11,13,11,
7,9,10,10,7,
7,12,6,9,15,
10,11,12,7,8,
8,9,11,10,9,
10,9,9,7,9,
11,7,10,10,11,
9,12,12,8,13,
9,10,8,11,10,
13,13,9,10,13
]

females = [
11,9,7,10,9,
10,10,7,9,10,
11,8,9,6,11,
10,7,9,12,14,
11,12,12,8,12,
12,9,10,11,7,
12,7,9,8,11,
10,8,13,8,10,
9,9,9,11,9,
9,8,9,12,11
]

# Combine datasets
data = males + females
data.sort()

n = len(data)

# Basic statistics
mean = statistics.mean(data)
median = statistics.median(data)
std = statistics.stdev(data)
var = statistics.variance(data)
minimum = min(data)
maximum = max(data)

# Quartile function
def percentile(p):
    k = (n-1) * p
    f = math.floor(k)
    c = math.ceil(k)
    if f == c:
        return data[int(k)]
    d0 = data[f] * (c-k)
    d1 = data[c] * (k-f)
    return d0+d1

# Quartiles
Q1 = percentile(0.25)
Q2 = percentile(0.50)
Q3 = percentile(0.75)

# Deciles
deciles = [percentile(i/10) for i in range(1,10)]

# Skewness
skew = sum((x-mean)**3 for x in data)/n
skew = skew/(std**3)

# Kurtosis
kurt = sum((x-mean)**4 for x in data)/n
kurt = kurt/(std**4) - 3

# 95% confidence interval
t = 1.984
margin = t*(std/math.sqrt(n))
ci_lower = mean - margin
ci_upper = mean + margin

print("Sample size:", n)
print("Mean:", mean)
print("Median:", median)
print("Standard Deviation:", std)
print("Variance:", var)
print("Min:", minimum)
print("Max:", maximum)
print("Skewness:", skew)
print("Kurtosis:", kurt)

print("Quartiles:")
print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)

print("Deciles:")
for i,d in enumerate(deciles,1):
    print("D"+str(i)+":", d)

print("95% CI:", (ci_lower, ci_upper))
```
