```python
import statistics
import math

print("Question 2:")
# Male and female data
males = [
12,4,11,13,11,
7,9,10,10,7,
7,12,6,9,15,
10,11,12,7,8,
8,9,11,10,9,
10,9,9,7,9,
11,7,10,10,11,
9,12,12,8,13,
9,10,8,11,10,
13,13,9,10,13
]

females = [
11,9,7,10,9,
10,10,7,9,10,
11,8,9,6,11,
10,7,9,12,14,
11,12,12,8,12,
12,9,10,11,7,
12,7,9,8,11,
10,8,13,8,10,
9,9,9,11,9,
9,8,9,12,11
]

def descriptive_stats(data, t_value=2.009):
    data_sorted = sorted(data)
    n = len(data)
    mean = statistics.mean(data)
    median = statistics.median(data)
    std = statistics.stdev(data)
    var = statistics.variance(data)
    minimum = min(data)
    maximum = max(data)
    
    # Skewness
    skew = sum((x-mean)**3 for x in data)/n
    skew = skew/(std**3)
    
    # Kurtosis
    kurt = sum((x-mean)**4 for x in data)/n
    kurt = kurt/(std**4) - 3
    
    # Quartile helper
    def percentile(p):
        k = (n-1)*p
        f = math.floor(k)
        c = math.ceil(k)
        if f == c:
            return data_sorted[int(k)]
        d0 = data_sorted[f] * (c-k)
        d1 = data_sorted[c] * (k-f)
        return d0+d1
    
    # Quartiles
    Q1 = percentile(0.25)
    Q2 = percentile(0.50)
    Q3 = percentile(0.75)
    
    # Deciles
    deciles = [percentile(i/10) for i in range(1,10)]
    
    # 95% CI
    margin = t_value * (std / math.sqrt(n))
    ci_lower = mean - margin
    ci_upper = mean + margin
    
    return {
        "n": n,
        "mean": mean,
        "median": median,
        "std_dev": std,
        "variance": var,
        "min": minimum,
        "max": maximum,
        "skewness": skew,
        "kurtosis": kurt,
        "Q1": Q1,
        "Q2": Q2,
        "Q3": Q3,
        "deciles": deciles,
        "95% CI": (ci_lower, ci_upper)
    }

# t-value for 95% CI (approx for n=50)
t_value = 2.009

male_stats = descriptive_stats(males, t_value)
female_stats = descriptive_stats(females, t_value)

print("***** Male Statistics *****")
for k,v in male_stats.items():
    print(f"{k}: {v}")

print("\n***** Female Statistics *****")
for k,v in female_stats.items():
    print(f"{k}: {v}")
```
