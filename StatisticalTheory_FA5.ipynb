```python
import itertools
from collections import Counter
import math
print("Problem 1")
print("a. The probability that there is no king on the first draw AND no king on the second draw.")
print("b. The probability that there is a king on the first draw OR on the second draw (or both) ")
print("c. The probability that the third card is a king GIVEN that the first and second cards were both kings.")
print("d. The event that there is no king on the first draw, no king on the second draw, but a king on the third draw.")



print("\n")
print("Problem 2")
print("a.)")
# All possible outcomes of rolling 3 dice
outcomes = list(itertools.product(range(1, 7), repeat=3))

# Compute sums
sums = [sum(o) for o in outcomes]

# Count frequencies
freq = Counter(sums)

# Total outcomes
total = len(outcomes)

# Display probability distribution
for s in sorted(freq):
    print(f"sums = {s}, Frequency = {freq[s]}, Probability = {freq[s]/total}") 

print("b.)")
print("15+21+25+27+27=115/216")
# Compute probability
prob = sum(freq[s] for s in range(7, 12)) / total
print("P(7 ≤ X ≤ 11) =", prob)
print("\n")

print("Problem 3")

# Values of X
X = [-10, -20, 30]

# Corresponding probabilities
P = [1/5, 3/10, 1/2]

# (a) E(X)
E_X = sum(x*p for x, p in zip(X, P))
print("E(X) =", E_X)

# (b) E(X^2)
E_X2 = sum((x**2)*p for x, p in zip(X, P))
print("E(X^2) =", E_X2)

# (c) E[(X - E(X))^2]  (Variance)
Var_X = sum(((x - E_X)**2)*p for x, p in zip(X, P))
print("E[(X - E(X))^2] =", Var_X)

# (d) E(X^3)
E_X3 = sum((x**3)*p for x, p in zip(X, P))
print("E(X^3) =", E_X3)
print("\n")

print("Problem 4")

# (a) No restriction
a = math.factorial(5)   # (6-1)!
print("(a.) No restriction =", a)

# (b) Two particular women must not sit together

# Total arrangements
total = math.factorial(5)

# Arrangements where the two particular women sit together
# Treat them as one block → 5 people total
together = math.factorial(4) * 2

b = total - together
print("(b.) Two particular women not together =", b)

# (c) Each woman must be between two men

# Arrange men first (circular)
men_arrange = math.factorial(2)

# Women fill the 3 spaces between men
women_arrange = math.factorial(3)

c = men_arrange * women_arrange
print("(c.) Each woman between two men =", c)
print("\n")

print("Problem 5")

# Combination function
def nCr(n, r):
    return math.comb(n, r)

# (a) No restriction
a = nCr(6,2) * nCr(8,4) * nCr(4,3) * nCr(5,3)
print("(a) No restriction =", a)

# (b) Particular man and woman must be selected
b = nCr(5,1) * nCr(7,3) * nCr(4,3) * nCr(5,3)
print("(b) Particular man and woman selected =", b)
print("\n")

print("Problem 6")
P_E1 = 0.6
P_E2 = 0.5
P_intersection = 0.3   # must be <= both

left_side = P_E1 + P_E2 - P_intersection
right_side = P_E1 + P_E2

print("P(E1 + E2) =", left_side)
print("P(E1) + P(E2) =", right_side)
print("Inequality holds?", left_side <= right_side)
```
