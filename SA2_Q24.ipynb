```python
# ==========================================
# ALZHEIMER'S MICE DATASET
# 2-WAY ANOVA USING GOOGLE COLAB
# ==========================================

# Install libraries if needed
!pip install pandas statsmodels scipy

# Import libraries
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
from scipy.stats import shapiro, levene

# ==========================================
# CREATE DATAFRAME
# ==========================================

data = [
    [1,1,12,10],
    [1,1,15,12],
    [1,1,13,13],
    [1,1,12,10],
    [1,1,14,13],

    [1,2,15,13],
    [1,2,17,13],
    [1,2,16,14],
    [1,2,17,15],
    [1,2,14,11],

    [1,3,13,12],
    [1,3,14,11],
    [1,3,18,15],
    [1,3,15,10],
    [1,3,16,14],

    [1,4,14,12],
    [1,4,13,11],
    [1,4,12,10],
    [1,4,14,13],
    [1,4,15,10],

    [2,1,17,9],
    [2,1,16,8],
    [2,1,17,10],
    [2,1,14,8],
    [2,1,13,8],

    [2,2,14,7],
    [2,2,18,10],
    [2,2,16,5],
    [2,2,17,9],
    [2,2,14,7],

    [2,3,13,8],
    [2,3,14,7],
    [2,3,18,9],
    [2,3,15,8],
    [2,3,16,9],

    [2,4,14,7],
    [2,4,13,9],
    [2,4,12,5],
    [2,4,14,8],
    [2,4,15,4]
]

columns = ['AD_Status', 'Treatment', 'Training', 'Memory']

df = pd.DataFrame(data, columns=columns)

print("DATASET")
print(df)

# ==========================================
# TRAINING DAY ANOVA
# ==========================================

print("\n==========================================")
print("TRAINING DAY 2-WAY ANOVA")
print("==========================================")

training_model = ols(
    'Training ~ C(AD_Status) * C(Treatment)',
    data=df
).fit()

training_anova = sm.stats.anova_lm(training_model, typ=2)

print("\nANOVA TABLE")
print(training_anova)

# ==========================================
# ASSUMPTION CHECKS - TRAINING
# ==========================================

print("\nNORMALITY TEST (Shapiro-Wilk)")
shapiro_training = shapiro(training_model.resid)
print(shapiro_training)

print("\nHOMOGENEITY OF VARIANCE (Levene Test)")
levene_training = levene(
    *[
        group["Training"].values
        for name, group in df.groupby(["AD_Status", "Treatment"])
    ]
)
print(levene_training)

# ==========================================
# MEMORY DAY ANOVA
# ==========================================

print("\n==========================================")
print("MEMORY DAY 2-WAY ANOVA")
print("==========================================")

memory_model = ols(
    'Memory ~ C(AD_Status) * C(Treatment)',
    data=df
).fit()

memory_anova = sm.stats.anova_lm(memory_model, typ=2)

print("\nANOVA TABLE")
print(memory_anova)

# ==========================================
# ASSUMPTION CHECKS - MEMORY
# ==========================================

print("\nNORMALITY TEST (Shapiro-Wilk)")
shapiro_memory = shapiro(memory_model.resid)
print(shapiro_memory)

print("\nHOMOGENEITY OF VARIANCE (Levene Test)")
levene_memory = levene(
    *[
        group["Memory"].values
        for name, group in df.groupby(["AD_Status", "Treatment"])
    ]
)
print(levene_memory)

# ==========================================
# APA STYLE INTERPRETATION
# ==========================================

print("\n==========================================")
print("APA STYLE INTERPRETATION")
print("==========================================")

print("""
TRAINING DAY:

A two-way ANOVA was conducted to determine the effects
of AD status and treatment on training day maze errors.

Interpret the p-values:
- p < .05 = significant
- p > .05 = not significant

MEMORY DAY:

A two-way ANOVA was conducted to determine the effects
of AD status and treatment on memory day maze errors.

Interpret the p-values:
- p < .05 = significant
- p > .05 = not significant
""")
```
